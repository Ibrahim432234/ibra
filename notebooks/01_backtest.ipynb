# Backtesting-Notebook (Demo)

Dieses Notebook ist ein Platzhalter für erste Walk-Forward-Backtests.

## Schritte

1. Lade das konfigurierte Dataset (`data/processed/features.parquet`).
2. Implementiere eine Walk-Forward-Validierung mit realistischer Kostenstruktur.
3. Berechne Metriken wie Profit Factor, Sharpe Ratio, Max Drawdown.
4. Grafische Auswertung (Equity-Kurve, Konfusionsmatrix, Feature-Importance).

> **Hinweis:** Dieses Notebook ist bewusst minimal gehalten und soll als Startpunkt dienen.
